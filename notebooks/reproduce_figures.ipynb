# Figures 2-4

```python
import sys
from pathlib import Path

ROOT = Path("..").resolve()
sys.path.insert(0, str(ROOT / "src"))

import style
import figure2
import figure3
import figure4

DATA = ROOT / "data"
OUT = ROOT / "outputs"
(OUT / "tables").mkdir(parents=True, exist_ok=True)
style.setup()


def write_tables(name, tables):
    for key, table in tables.items():
        table.to_csv(OUT / "tables" / f"{name}_{key}.csv", index=False)
```

## Figure 2

```python
tables2 = figure2.tables(DATA / "figure2")
write_tables("figure2", tables2)
```

```python
fig2 = figure2.draw(tables2, DATA / "figure2")
style.save(fig2, OUT / "figure2")
```

## Figure 3

```python
tables3 = figure3.tables(DATA / "figure3")
write_tables("figure3", tables3)
```

```python
fig3 = figure3.draw(tables3, DATA / "figure3")
style.save(fig3, OUT / "figure3")
```

```python
tables4 = figure4.tables(DATA / "figure4")
write_tables("figure4", tables4)
```
